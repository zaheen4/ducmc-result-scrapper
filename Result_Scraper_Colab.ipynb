```python
# Cell 1: Mount Google Drive
from google.colab import drive
import os

if not os.path.exists('/content/drive/MyDrive'):
    drive.mount('/content/drive')
else:
    print('Drive already mounted')
```

```python
# Cell 2: Edit config.json (run before scraper to change settings)
import json

DATA_DIR = '/content/drive/MyDrive/ResultScraperData'
CONFIG_PATH = os.path.join(DATA_DIR, 'config.json')

config = {}
if os.path.exists(CONFIG_PATH):
    with open(CONFIG_PATH) as f:
        config = json.load(f)

def _input(prompt, default):
    hint = f' [{default}]' if default else ''
    val = input(f'{prompt}{hint}: ').strip()
    return val if val else str(default)

print('Press Enter to keep current value (shown in brackets).\n')
config['google_sheet_url'] = _input('Google Sheet URL', config.get('google_sheet_url', ''))
config['worksheet_name'] = _input('Worksheet name', config.get('worksheet_name', ''))
config['program'] = _input('Program name', config.get('program', 'B.Sc. in Computer Science and Engineering'))
config['session'] = _input('Session (e.g. 2021-2022)', config.get('session', '2021-2022'))
config['retake_exam'] = _input('Retake exam (code like L1T2R-2024, empty to select interactively)', config.get('retake_exam', ''))
config['start_regi'] = int(_input('Start regi number', config.get('start_regi', 710)))
config['end_regi'] = int(_input('End regi number', config.get('end_regi', 813)))
config['request_delay'] = int(_input('Seconds between requests', config.get('request_delay', 1)))

with open(CONFIG_PATH, 'w') as f:
    json.dump(config, f, indent=2, ensure_ascii=False)
print(f'\nConfig saved to {CONFIG_PATH}')
```

```python
# Cell 3: Copy scraper files to Colab workspace
import shutil

DRIVE_DIR = '/content/drive/MyDrive/ResultScraperData'
WORKSPACE = '/content'

for f in ['scraper_common.py', 'colab_scrapper.py', 'credentials.json',
          'targets.json', 'exam_catalog.json', 'cse_credit_hours.json']:
    src = os.path.join(DRIVE_DIR, f)
    dst = os.path.join(WORKSPACE, f)
    if os.path.exists(src):
        shutil.copy2(src, dst)
        print(f'Copied {f}')
    else:
        print(f'MISSING: {f} — upload it to {DRIVE_DIR}')
```
